```python
import numpy as np

# Remove completely empty column
df = df.drop(columns=["_c39"])

# Convert '?' to NaN
df = df.replace("?", np.nan)

# Check missing values again
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
numeric_features = [
    "age",
    "months_as_customer",
    "policy_deductable",
    "policy_annual_premium",
    "number_of_vehicles_involved",
    "bodily_injuries",
    "witnesses",
    "incident_hour_of_the_day"
]

categorical_features = [
    "incident_type",
    "collision_type",
    "incident_severity",
    "property_damage",
    "police_report_available"
]

target = "total_claim_amount"

X = df[numeric_features + categorical_features]
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.25,
    random_state=42
)

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])
```

```python
print(preprocessor)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

linear_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])
```

```python
linear_model.fit(X_train, y_train)
```

```python
y_train_pred = linear_model.predict(X_train)
y_val_pred = linear_model.predict(X_val)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

train_mae = mean_absolute_error(y_train, y_train_pred)
train_mse = mean_squared_error(y_train, y_train_pred)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(y_train, y_train_pred)

val_mae = mean_absolute_error(y_val, y_val_pred)
val_mse = mean_squared_error(y_val, y_val_pred)
val_rmse = np.sqrt(val_mse)
val_r2 = r2_score(y_val, y_val_pred)

print("===== TRAINING =====")
print("MAE :", train_mae)
print("MSE :", train_mse)
print("RMSE:", train_rmse)
print("R²  :", train_r2)

print("\n===== VALIDATION =====")
print("MAE :", val_mae)
print("MSE :", val_mse)
print("RMSE:", val_rmse)
print("R²  :", val_r2)
```

```python
from sklearn.linear_model import Ridge

ridge_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=1.0))
])
```

```python
ridge_model.fit(X_train, y_train)
```

```python
y_train_pred_ridge = ridge_model.predict(X_train)
y_val_pred_ridge = ridge_model.predict(X_val)
```

```python
train_mae = mean_absolute_error(y_train, y_train_pred_ridge)
train_mse = mean_squared_error(y_train, y_train_pred_ridge)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(y_train, y_train_pred_ridge)

val_mae = mean_absolute_error(y_val, y_val_pred_ridge)
val_mse = mean_squared_error(y_val, y_val_pred_ridge)
val_rmse = np.sqrt(val_mse)
val_r2 = r2_score(y_val, y_val_pred_ridge)

print("===== RIDGE TRAINING =====")
print("MAE :", train_mae)
print("MSE :", train_mse)
print("RMSE:", train_rmse)
print("R²  :", train_r2)

print("\n===== RIDGE VALIDATION =====")
print("MAE :", val_mae)
print("MSE :", val_mse)
print("RMSE:", val_rmse)
print("R²  :", val_r2)
```

```python
alphas = [0.01, 0.1, 1, 10, 100]

for alpha in alphas:

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("model", Ridge(alpha=alpha))
    ])

    model.fit(X_train, y_train)

    y_val_pred = model.predict(X_val)

    mae = mean_absolute_error(y_val, y_val_pred)
    mse = mean_squared_error(y_val, y_val_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_val, y_val_pred)

    print(f"Alpha: {alpha}")
    print(f"MAE : {mae:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"R²  : {r2:.4f}")
    print("-" * 30)
```

```python
from sklearn.model_selection import cross_val_score

alphas = [0.01, 0.1, 1, 10, 100]

for alpha in alphas:
  
    model = Pipeline([
        ("preprocessor", preprocessor),
        ("model", Ridge(alpha=alpha))
    ])

    scores = cross_val_score(
        model,
        X_train_val,
        y_train_val,
        cv=5,
        scoring="r2"
    )

    print(f"Alpha: {alpha}")
    print("R² scores:", scores)
    print("Mean R²:", scores.mean())
    print("Std R² :", scores.std())
    print("-" * 40)
```

```python
final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=0.1))
])
```

```python
final_model.fit(X_train_val, y_train_val)
```

```python
y_test_pred = final_model.predict(X_test)
```

```python
test_mae = mean_absolute_error(y_test, y_test_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
test_rmse = np.sqrt(test_mse)
test_r2 = r2_score(y_test, y_test_pred)

print("===== FINAL TEST RESULTS =====")
print("MAE :", test_mae)
print("MSE :", test_mse)
print("RMSE:", test_rmse)
print("R²  :", test_r2)
```

```python
feature_names = final_model.named_steps["preprocessor"].get_feature_names_out()

coefficients = final_model.named_steps["model"].coef_

coef_df = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefficients
})

coef_df["abs_coefficient"] = coef_df["coefficient"].abs()

coef_df = coef_df.sort_values(
    "abs_coefficient",
    ascending=False
)

print(coef_df.head(15))
```

```python
import joblib

joblib.dump(final_model, "insurance_ridge_model.pkl")

print("Model saved successfully!")
```
